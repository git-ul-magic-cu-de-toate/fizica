```python
#set time segment
time=0.05
```

```python
#create time array
time_array=np.linspace(0.0,285*time, 285)
```

```python
import scipy.integrate
```

```python
#calculate time differential
delta_t=np.mean(np.diff(time_array))
```

```python
#convert acceleration to numpy array 
x_acceleration=df["pos_x"].to_numpy()
y_acceleration=df["pos_y"].to_numpy()
z_acceleration=df["pos_z"].to_numpy()
```

```python
from scipy import signal
```

```python
#detrend values for acceleration
x_acceleration=signal.detrend(x_acceleration)
y_acceleration=signal.detrend(y_acceleration)
z_acceleration=signal.detrend(z_acceleration)
```

```python
#find velocity given acceleration
x_speed=scipy.integrate.cumtrapz(x_acceleration, dx = delta_t, initial = 0) 
y_speed=scipy.integrate.cumtrapz(y_acceleration, dx = delta_t, initial = 0) 
z_speed=scipy.integrate.cumtrapz(z_acceleration, dx = delta_t, initial = 0) 
```

```python
#detrend velocity values
x_speed=signal.detrend(x_speed)
y_speed=signal.detrend(y_speed)
z_speed=signal.detrend(z_speed)
```
